1. order_approved_at: Has some null values (e.g., for rows with status 'canceled', 'processing', and 'shipped').

2. order_delivered_carrier_date: Has numerous null values. This usually indicates that the order was never picked up by the carrier. This can be due to 'canceled', 'unavailable', or 'shipped' (but not yet delivered) statuses.

3. order_delivered_customer_date: Has many null values. This means the order was never delivered to the customer. This is common for orders with statuses like 'shipped', 'processing', 'invoiced', and 'canceled'.

```python
# can be filled with the median:

numeric_col = [
   'product_name_lenght',
    'product_description_lenght',
    'product_photos_qty',
    'product_weight_g',
    'product_length_cm',
    'product_height_cm',
    'product_width_cm'
]

for col in numeric_col:
    products[col] = products[col].fillna(products[col].median())
```

Why use the median?

For numeric columns such as weight and dimensions, the median is preferred because it is less affected by very large or very small values than the mean.

# Now we are creating a new Master DataFrame 

```python
import pandas as pd

customers = pd.read_csv("customers_clean.csv")
orders = pd.read_csv("orders_clean.csv")
payments = pd.read_csv("payments_clean.csv")
order_items = pd.read_csv("order_items_clean.csv")
products = pd.read_csv("products_clean.csv")
reviews = pd.read_csv("order_reviews_clean.csv")
sellers = pd.read_csv("sellers_clean.csv")
category_translation = pd.read_csv("category_translation_clean.csv")
```

We are not merging the geolocation dataset because it doesn't have a direct key with the other tables. It can be analyzed separately later.

```python
master_df = orders.merge(customers,on='customer_id',how='left')
master_df = master_df.merge(order_items,on='order_id',how='left')
master_df = master_df.merge(payments,on='order_id',how='left')
master_df = master_df.merge(products,on='product_id',how='left')
master_df = master_df.merge(sellers,on='seller_id',how='left')
master_df = master_df.merge(reviews,on='order_id',how='left')
master_df = master_df.merge(category_translation,on='product_category_name',how='left')
```

```python
master_df.info()
```

```python
master_df.isnull().sum()
```

```python
missing = master_df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
missing
```

Some missing values are expected because:

- Cancelled orders
- Orders without reviews
- Unknown product categories

Do not remove them without understanding why they are missing.
